```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
age = np.array([18,20,22,24,26,28,30]).reshape(-1,1)
acci = np.array([37,41,45,49,53,57,61])
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
model.fit(age,acci)
```

```python
value = model.predict([[80]])
```

```python
print(value[0])
```

```python
co = model.coef_[0]
yin = model.intercept_
```

```python
print(co,yin)
```

```python
plt.scatter(x=age,y=acci)
plt.plot(age,model.predict(age))
```
