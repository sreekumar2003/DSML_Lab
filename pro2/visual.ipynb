```python

iris = sns.load_dataset('iris')
```

```python

plt.hist(iris['petal_length'],color='skyblue', edgecolor='black')
plt.title('Histogram of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.show()

plt.hist(iris['petal_length'],color='skyblue', edgecolor='black')
plt.title('Histogram of Sepal Length')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.show()

plt.hist(iris['petal_width'],color='skyblue', edgecolor='black')
plt.title('Histogram of Sepal Length')
plt.xlabel('petal_width(cm)')
plt.ylabel('Frequency')
plt.show()

plt.hist(iris['sepal_width'],color='skyblue', edgecolor='black')
plt.title('Histogram of Sepal Length')
plt.xlabel('sepal_width(cm)')
plt.ylabel('Frequency')
plt.show()


```

```python

sns.countplot(x='species', data=iris,color='yellow')


plt.title('Countplot of Iris Species')
plt.xlabel('Species')
plt.ylabel('Count')


plt.show()
```
