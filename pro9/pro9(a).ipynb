```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
car_age = np.array([5, 7, 8, 7, 2, 17, 2, 9, 4, 11, 12, 9, 6]).reshape(-1, 1)  # Reshape for sklearn
car_speed = np.array([99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86])
```

```python
model = LinearRegression()
model.fit(car_age, car_speed)
```

```python
# Step 2: Get the slope (m) and intercept (b)
m = model.coef_[0]  # Slope
b = model.intercept_  # Intercept
```

```python
car_age_14 = 14
predicted_speed_14 = model.predict([[car_age_14]]) 
```

```python
plt.scatter(car_age, car_speed, color='blue', label='Data Points')
plt.plot(car_age, model.predict(car_age), color='red', label=f'Linear Regression: y = {m:.2f}x + {b:.2f}')
plt.xlabel('Car Age (years)')
plt.ylabel('Car Speed (km/h)')
plt.title('Car Age vs. Car Speed with Linear Regression Line')
plt.legend()
plt.show()
```

```python
# Output the results
print(f"Slope (m): {m}")
print(f"Intercept (b): {b}")
print(f"Predicted Car Speed for Car Age 14: {predicted_speed_14[0]}")
```
