```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
age = np.array([18, 22, 30, 45, 65, 80]).reshape(-1, 1)  # Reshape for sklearn
accident_no = np.array([38, 36, 24, 20, 18, 28])
```

```python
model = LinearRegression()
model.fit(age, accident_no)
```

```python
# Step 2: Get the slope (m) and intercept (b)
m = model.coef_[0]  # Slope
b = model.intercept_  # Intercept
```

```python
age_40 = 40
age_60 = 60
predicted_accidents_40 = model.predict([[age_40]])  # Sklearn expects 2D array
predicted_accidents_60 = model.predict([[age_60]])
```

```python
plt.scatter(age, accident_no, color='blue', label='Data Points')
plt.plot(age, model.predict(age), color='red', label=f'Linear Regression: y = {m:.2f}x + {b:.2f}')
plt.xlabel('Age')
plt.ylabel('Number of Accidents')
plt.title('Age vs Number of Accidents with Linear Regression Line')
plt.legend()
plt.show()
```

```python
print(f"Slope (m): {m}")
print(f"Intercept (b): {b}")
print(f"Predicted Number of Accidents for Age 40: {predicted_accidents_40[0]}")
print(f"Predicted Number of Accidents for Age 60: {predicted_accidents_60[0]}")
```
