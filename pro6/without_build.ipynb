```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
```

```python
iris = load_iris()
X = iris.data
y = iris.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
def euclidean_distance(a, b):
    return np.sqrt(np.sum((a - b) ** 2))

def get_neighbors(X_train, y_train, test_instance, k):
    distances = []
    for i in range(len(X_train)):
        dist = euclidean_distance(X_train[i], test_instance)
        distances.append((y_train[i], dist))
    distances.sort(key=lambda x: x[1])  # sort by distance
    neighbors = distances[:k]
    return neighbors
```

```python
# 5. Predict function using majority vote
def predict(X_train, y_train, test_instance, k):
    neighbors = get_neighbors(X_train, y_train, test_instance, k)
    # Count the classes of neighbors
    classes = [neighbor[0] for neighbor in neighbors]
    prediction = max(set(classes), key=classes.count)  # majority vote
    return prediction
```

```python

k = 5
predictions = []

for test_instance in X_test:
    pred = predict(X_train, y_train, test_instance, k)
    predictions.append(pred)

accuracy = np.sum(predictions == y_test) / len(y_test)
print(f'Accuracy: {accuracy * 100:.2f}%')

my_sample = np.array([5.1, 3.5, 1.4, 0.2])

predicted_class = predict(X_train, y_train, my_sample, k=5)

species = iris.target_names[predicted_class]

print(f'The predicted class for your input is: {species}')
```
