```python
import cv2

img = cv2.imread('sample.jpg')

# Resize to half the original size
height, width = img.shape[:2]
resized = cv2.resize(img, (width//2, height//2), interpolation=cv2.INTER_LINEAR)

# Crop a region (e.g., top-left quarter)
cropped = img[0:height//2, 0:width//2]

# Save results
cv2.imwrite('resized.jpg', resized)
cv2.imwrite('cropped.jpg', cropped)
print("Resized and cropped images saved.")
```
