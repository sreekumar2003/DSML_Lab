```python
# Import necessary Bokeh libraries
from bokeh.plotting import figure, show, output_notebook
import numpy as np
```

```python
# Set up Bokeh to display plots in the notebook
output_notebook()
```

```python
# Create some data for the plot
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create a figure object
p = figure(title="Simple Line Plot", x_axis_label='X', y_axis_label='Y')

# Add a line renderer with the data
p.line(x, y, legend_label="sin(x)", line_width=2)

# Show the plot in the notebook
show(p)
```

```python
from bokeh.models import HoverTool

# Generate random data
x = np.random.rand(50) * 10
y = np.random.rand(50) * 10
sizes = np.random.randint(5, 50, 50)

# Create the figure object
p = figure(title="Scatter Plot with Hover", x_axis_label='X', y_axis_label='Y')

# Add a circle renderer (scatter plot)
p.circle(x, y, size=sizes, color="navy", alpha=0.5)

# Create a HoverTool to display the values of x, y, and size
hover = HoverTool()
hover.tooltips = [("X", "$x"), ("Y", "$y"), ("Size", "@size")]

# Add the hover tool to the plot
p.add_tools(hover)

# Show the plot in the notebook
show(p)
```

```python
from bokeh.layouts import column
from bokeh.models import Slider
from bokeh.io import curdoc

# Create data for the plot
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Create the figure object
p = figure(title="Interactive Sine Wave", x_axis_label='X', y_axis_label='Y')

# Add a line renderer
line = p.line(x, y, line_width=2, legend_label="sin(x)")

# Create a slider to adjust frequency
slider = Slider(start=1, end=10, value=1, step=0.1, title="Frequency")

# Callback function to update the sine wave when the slider value changes
def update(attr, old, new):
    freq = slider.value
    y_new = np.sin(freq * x)  # Update sine wave with new frequency
    line.data_source.data['y'] = y_new  # Update y data in the plot

# Attach the update function to the slider's value change
slider.on_change('value', update)

# Combine the slider and plot into a layout
layout = column(slider, p)

# Show the interactive plot in the notebook
show(layout)
```

```python
from bokeh.plotting import output_file

# Set output file for saving the plot
output_file("plot.html")

# Show the plot (it will save it as an HTML file)
show(p)
```
