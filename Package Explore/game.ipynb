```python
import pygame
import random

# Initialize Pygame
pygame.init()

# Game window dimensions
screen_width = 800
screen_height = 600
screen = pygame.display.set_mode((screen_width, screen_height))
pygame.display.set_caption("Dodge the Falling Blocks")

# Define colors
WHITE = (255, 255, 255)
RED = (255, 0, 0)
BLUE = (0, 0, 255)

# Player properties
player_width = 50
player_height = 50
player_x = screen_width // 2
player_y = screen_height - player_height - 10
player_speed = 10

# Falling block properties
block_width = 50
block_height = 50
block_speed = 5

# Create a clock object to control FPS
clock = pygame.time.Clock()

# Game loop
running = True
blocks = []
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    # Get the pressed keys
    keys = pygame.key.get_pressed()

    # Move the player
    if keys[pygame.K_LEFT] and player_x > 0:
        player_x -= player_speed
    if keys[pygame.K_RIGHT] and player_x < screen_width - player_width:
        player_x += player_speed

    # Create new falling blocks
    if random.randint(1, 100) <= 2:  # 2% chance to create a block each frame
        block_x = random.randint(0, screen_width - block_width)
        blocks.append([block_x, 0])  # Start at the top of the screen

    # Move the blocks
    for block in blocks:
        block[1] += block_speed  # Move down

    # Check for collisions (if block reaches the player)
    for block in blocks:
        if block[1] > player_y and block[1] < player_y + player_height and \
           block[0] > player_x and block[0] < player_x + player_width:
            running = False  # End the game if collision happens

    # Remove blocks that have fallen off screen
    blocks = [block for block in blocks if block[1] < screen_height]

    # Fill the screen with white
    screen.fill(WHITE)

    # Draw the player
    pygame.draw.rect(screen, BLUE, (player_x, player_y, player_width, player_height))

    # Draw the falling blocks
    for block in blocks:
        pygame.draw.rect(screen, RED, (block[0], block[1], block_width, block_height))

    # Update the display
    pygame.display.update()

    # Set FPS
    clock.tick(60)

# Quit Pygame
pygame.quit()
```
