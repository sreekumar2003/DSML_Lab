```python
import numpy as np
from scipy import optimize

# Define the function
def func(x):
    return x**2 + 4*x + 4

# Use scipy's minimize function to find the minimum
result = optimize.minimize(func, x0=0)  # Starting guess for x is 0

print(f"Minimum value: {result.fun}")
print(f"Value of x at the minimum: {result.x}")
```

```python
from scipy import integrate

# Define the function
def integrand(x):
    return x**2

# Compute the definite integral of x^2 from 0 to 1
result, error = integrate.quad(integrand, 0, 1)

print(f"Integral result: {result}")
print(f"Estimated error: {error}")
```

```python
from scipy import interpolate
import matplotlib.pyplot as plt

# Define some data points
x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 1, 4, 9, 16])

# Create a linear interpolator
f = interpolate.interp1d(x, y, kind='linear')

# Generate new points for interpolation
x_new = np.linspace(0, 4, 10)
y_new = f(x_new)

# Plot the original data and the interpolation
plt.plot(x, y, 'o', label='Original Data')
plt.plot(x_new, y_new, '-', label='Linear Interpolation')
plt.legend()
plt.show()
```

```python
from scipy import linalg

# Define the coefficient matrix and the right-hand side vector
A = np.array([[2, 1], [3, 2]])
b = np.array([1, 2])

# Solve the system of linear equations Ax = b
solution = linalg.solve(A, b)

print(f"Solution: x = {solution}")
```

```python
from scipy.optimize import minimize

# Define the function to minimize
def func(x):
    return x**2

# Define the constraint: x >= 1
constraint = {'type': 'ineq', 'fun': lambda x: x - 1}

# Minimize the function with the constraint
result = minimize(func, 0, constraints=constraint)

print(f"Optimized result: {result.x}")
```
